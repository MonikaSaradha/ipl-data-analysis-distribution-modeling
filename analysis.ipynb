# IPL Data Analysis & Distribution Modeling

Analysis of IPL datasets to study player performance, statistical distributions, and salary relationships.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

## Load Datasets

```python
ball = pd.read_excel('ipl_ball_by_ball.xlsx')
salary = pd.read_excel('ipl_salary_data.xlsx')
match = pd.read_csv('ipl_matches.csv')
```

## Data Preprocessing

```python
print(ball.info())
print(match.isna().sum())

match = match.dropna()
```

## Top Run Getters

```python
top_runs = ball.groupby(['Season','batter'])['batsman_run'].sum().reset_index()
top_runs = top_runs.sort_values(['Season','batsman_run'], ascending=[True, False])

top_runs.head()
```

## Wicket Analysis

```python
wickets = ball.groupby(['Season','bowler'])['isWicketDelivery'].sum().reset_index()

wickets.head()
```

## Recent Seasons Analysis

```python
recent = ball.sort_values('Season', ascending=False).head(10000)

recent.head()
```

## Distribution Fitting (Runs)

```python
runs = recent['batsman_run']

params = stats.alpha.fit(runs)
ks_stat, p_val = stats.kstest(runs, 'alpha', args=params)

print('KS Test p-value (Runs):', p_val)
```

## Distribution Fitting (Wickets)

```python
wkts = recent['isWicketDelivery']

params = stats.alpha.fit(wkts)
ks_stat, p_val = stats.kstest(wkts, 'alpha', args=params)

print('KS Test p-value (Wickets):', p_val)
```

## Salary vs Performance

```python
corr = salary[['Runs','Matches','Sixes','Final Price']].corr()

print(corr)
```

```python
plt.scatter(salary['Runs'], salary['Final Price'])
plt.xlabel('Runs')
plt.ylabel('Salary')
plt.title('Runs vs Salary')
plt.show()
```

## Key Insights

- Top players dominate scoring across seasons
- Alpha distribution reasonably fits runs and wickets
- Salary is influenced by performance metrics like runs and matches
